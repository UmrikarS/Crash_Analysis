```python
from shapely.geometry import Point
df_gold = spark.table("gold_crashes_features").toPandas()
print(f"Loaded: {len(df_gold):,} crash records")

# WGS84 — X=longitude (~174.x), Y=latitude (~-36.x)
geometry = [Point(xy) for xy in zip(df_gold['X'], df_gold['Y'])]
gdf = gpd.GeoDataFrame(df_gold, geometry=geometry, crs="EPSG:4326")

print(f"CRS    : {gdf.crs}")
print(f"Bounds : {gdf.total_bounds}")
print(f"X range: {gdf['X'].min():.4f} – {gdf['X'].max():.4f}  (longitude)")
print(f"Y range: {gdf['Y'].min():.4f} – {gdf['Y'].max():.4f}  (latitude)")

gdf_severe = gdf[gdf['severe_crash'] == 1].copy()
print(f"\nTotal crashes : {len(gdf):,}")
print(f"Severe crashes: {len(gdf_severe):,}  ({len(gdf_severe)/len(gdf)*100:.1f}%)")
```

```python
import numpy as np
from sklearn.neighbors import KernelDensity

coords = np.array(list(zip(gdf_severe['X'], gdf_severe['Y'])))
n, d   = len(coords), 2

# ── Verify data bounds ────────────────────────────────────────────
print(f"Input: {n:,} severe crashes")
print(f"X (lon): {coords[:,0].min():.4f} – {coords[:,0].max():.4f}")
print(f"Y (lat): {coords[:,1].min():.4f} – {coords[:,1].max():.4f}")

# ── Correct Scott's rule for 2D ───────────────────────────────────
# h = n^(-1/(d+4)) × std — applied using mean std of lon and lat
# Pure Scott gives ~13km which blurs whole Auckland into one blob
# Scale ×0.5 for suburb-level hotspot resolution (~5–8km)
std_lon      = coords[:,0].std()
std_lat      = coords[:,1].std()
scott_factor = n ** (-1.0 / (d + 4))
bandwidth    = round(scott_factor * np.mean([std_lon, std_lat]) * 0.5, 6)

print(f"\\nScott factor   : {scott_factor:.6f}")
print(f"Std lon/lat    : {std_lon:.4f} / {std_lat:.4f}")
print(f"Applied bw     : {bandwidth:.6f}° (~{bandwidth*111:.1f} km suburb-level)")

kde = KernelDensity(bandwidth=bandwidth, kernel='gaussian')
kde.fit(coords)
print(f"\\nKDE fitted | bandwidth={bandwidth:.6f}° (~{bandwidth*111:.2f} km)")

# ── Grid — use actual data bounds + 0.05° buffer ─────────────────
lon_min = max(174.40, coords[:,0].min() - 0.05)
lon_max = min(175.35, coords[:,0].max() + 0.05)
lat_min = max(-37.20, coords[:,1].min() - 0.05)
lat_max = min(-36.60, coords[:,1].max() + 0.05)

xx, yy = np.meshgrid(
    np.linspace(lon_min, lon_max, 200),
    np.linspace(lat_min, lat_max, 200)
)
print(f"Grid: {xx.shape} | lon {lon_min:.3f}–{lon_max:.3f} lat {lat_min:.3f}–{lat_max:.3f}")

log_dens       = kde.score_samples(np.c_[xx.ravel(), yy.ravel()])
densities      = np.exp(log_dens).reshape(xx.shape)
densities_norm = (densities - densities.min()) / (densities.max() - densities.min())

peak_idx = densities_norm.argmax()
print(f"\\nDensity surface | range 0–1")
print(f"   High-density (>0.5): {(densities_norm > 0.5).sum()} cells")
print(f"   Peak location: lon={xx.ravel()[peak_idx]:.4f} lat={yy.ravel()[peak_idx]:.4f}")
```

```python
# ── Find top 10 peaks with ≥5km geographic separation ────────────────────────
# Naive argsort gives all 10 peaks in same 2km area (Eden Terrace)
# Fix: after finding each peak, zero out 5km radius before finding next

densities_work = densities_norm.copy()   # working copy — will be modified
top10 = []

for rank in range(1, 11):
    # Find current global maximum
    flat_idx = densities_work.ravel().argmax()
    lon      = float(xx.ravel()[flat_idx])
    lat      = float(yy.ravel()[flat_idx])
    score    = float(densities_norm.ravel()[flat_idx])  # score from original

    # Suppress 5km radius around this peak before next search
    # 0.045° lat ≈ 5km, 0.056° lon ≈ 5km at Auckland latitude
    suppress = (
        (np.abs(xx - lon) < 0.056) &
        (np.abs(yy - lat) < 0.045)
    )
    densities_work[suppress] = 0.0

    # Count crashes within 500m radius of this peak
    nearby = gdf[
        (np.abs(gdf['X'] - lon) < 0.0055) &
        (np.abs(gdf['Y'] - lat) < 0.0045)
    ]

    top10.append({
        'rank':              rank,
        'latitude':          round(lat, 5),
        'longitude':         round(lon, 5),
        'density_score':     round(score, 4),
        'total_crashes':     len(nearby),
        'severe_crashes':    int(nearby['severe_crash'].sum())
                             if 'severe_crash' in nearby.columns else 0,
        'fatal_count':       int(nearby['fatalCount'].sum())
                             if 'fatalCount' in nearby.columns else 0,
        'serious_count':     int(nearby['seriousInjuryCount'].sum())
                             if 'seriousInjuryCount' in nearby.columns else 0,
        'crash_location':    nearby['crashLocation1'].mode()[0]
                             if len(nearby) > 0 and 'crashLocation1' in nearby.columns
                             else 'Unknown',
        'dominant_severity': nearby['crashSeverity_clean'].mode()[0]
                             if len(nearby) > 0 and 'crashSeverity_clean' in nearby.columns
                             else 'Unknown',
    })

top10_df = pd.DataFrame(top10)

print("\\nTOP 10 KDE HOTSPOTS (geographically spread, 500m counts)")
print("=" * 95)
print("  Rank       Lat        Lon   Score  Total  Severe  Fatal  Serious  Location")
print("-" * 95)
for _, r in top10_df.iterrows():
    print(f"  #{int(r['rank'])} lat={r['latitude']:.4f} lon={r['longitude']:.4f} "
          f"score={r['density_score']:.4f} crashes={int(r['total_crashes'])} "
          f"severe={int(r['severe_crashes'])} fatal={int(r['fatal_count'])} "
          f"serious={int(r['serious_count'])} loc={str(r['crash_location'])[:28]}")
print("=" * 95)

# Verify geographic spread
print("\\n📏 Min distance between hotspot pairs (target ≥5km):")
lats = top10_df['latitude'].values; lons = top10_df['longitude'].values
min_dist = 999
for i in range(len(top10_df)):
    for j in range(i+1, len(top10_df)):
        d = (((lats[i]-lats[j])*111)**2 + ((lons[i]-lons[j])*89)**2)**0.5
        if d < min_dist: min_dist = d
        if d < 4: print(f"  #{i+1}↔#{j+1} = {d:.1f}km — try increasing suppress radius to 0.07°")
status_sep = "" if min_dist >= 4 else "increase suppress radius"
print(f"  Minimum separation: {min_dist:.1f}km " + status_sep)

spark.createDataFrame(top10_df).write.format("delta").mode("overwrite") \
     .saveAsTable("gold_top10_kde_hotspots")
print("\\nSaved: gold_top10_kde_hotspots")
```

```python
# ── Step 1: Add road labels using crashLocation1 from actual data ────────────
# Reuses the same 500m radius from Cell 4 — no double-counting
print("Adding road names from crashLocation1...")
primary_roads   = []
secondary_roads = []

for _, row in top10_df.iterrows():
    lat, lon = row['latitude'], row['longitude']
    nearby = gdf[
        (np.abs(gdf['X'] - lon) < 0.0055) &
        (np.abs(gdf['Y'] - lat) < 0.0045)
    ]
    if len(nearby) > 0 and 'crashLocation1' in nearby.columns:
        r1 = nearby['crashLocation1'].dropna().value_counts()
        primary_roads.append(r1.index[0] if len(r1) > 0 else 'Unknown')
    else:
        primary_roads.append('Unknown')
    if len(nearby) > 0 and 'crashLocation2' in nearby.columns:
        r2 = nearby['crashLocation2'].dropna().value_counts()
        secondary_roads.append(r2.index[0] if len(r2) > 0 else '')
    else:
        secondary_roads.append('')
    print(f"  #{int(row['rank'])} ({lat:.4f},{lon:.4f}) -> {primary_roads[-1][:30]}")

top10_df['primary_road']   = primary_roads
top10_df['secondary_road'] = secondary_roads

# ── Step 2: Reverse geocode suburb (10 requests, ~15 sec) ────────────────────
print("\\nReverse geocoding suburb context...")
geolocator = Nominatim(user_agent="mbi908_v3", timeout=10)
suburbs = []

for _, row in top10_df.iterrows():
    try:
        loc  = geolocator.reverse(
            (row['latitude'], row['longitude']), language="en", zoom=14)
        addr = loc.raw.get('address', {}) if loc else {}
        s    = (addr.get('suburb') or addr.get('neighbourhood') or
                addr.get('village') or addr.get('city_district') or 'Auckland')
    except Exception:
        s = 'Auckland'
    suburbs.append(s)
    print(f"  #{int(row['rank'])} suburb: {s}")
    time.sleep(1.1)   # Nominatim 1 req/sec rate limit

top10_df['suburb'] = suburbs

# ── Step 3: Disambiguate repeated suburb names ────────────────────────────────
# If two hotspots share suburb, append road name so popups stay distinct
suburb_counts = top10_df['suburb'].value_counts()
top10_df['display_name'] = top10_df.apply(
    lambda r: (f"{r['suburb']} ({r['primary_road'][:15]})"
               if suburb_counts[r['suburb']] > 1 else r['suburb']),
    axis=1
)

print("\\nTOP 10 KDE HOTSPOTS — FINAL")
print("=" * 85)
print("#    Display Name                   Road                      Crashes   Severe")
print("-" * 85)
for _, r in top10_df.iterrows():
    print(f"  #{int(r['rank'])} {str(r['display_name'])[:25]} | "
          f"{str(r['primary_road'])[:22]} | "
          f"crashes={int(r['total_crashes']):,} severe={int(r['severe_crashes'])}")
print("=" * 85)

spark.createDataFrame(top10_df).write.format("delta").mode("overwrite").option("overwriteSchema", "true") \
     .saveAsTable("gold_top10_kde_hotspots")
print("\\nSaved: gold_top10_kde_hotspots (primary_road, suburb, display_name)")
```

```python
# ── Severity breakdown per top 10 KDE hotspot — FIXED for actual column values ─
# crashSeverity_clean values: 'Unknown', 'Minor', 'Serious', 'Fatal'
# (Unknown = Non-Injury / no injury recorded)

import numpy as np, pandas as pd

df_gold = spark.table("gold_crashes_features").toPandas()
top10   = spark.table("gold_top10_kde_hotspots").toPandas()

rows = []
for _, hs in top10.iterrows():
    nearby = df_gold[
        (np.abs(df_gold['X'] - hs['longitude']) < 0.0055) &
        (np.abs(df_gold['Y'] - hs['latitude'])  < 0.0045)
    ].copy()

    total = len(nearby)
    if total == 0:
        continue

    sev_pcts = (
        nearby['crashSeverity_clean']
        .fillna('Unknown')
        .value_counts(normalize=True) * 100
    ).round(1)

    rows.append({
        'rank':           int(hs['rank']),
        'crash_location': str(hs['primary_road']),
        'display_name':   str(hs.get('display_name', hs['primary_road'])),
        'total_crashes':  total,
        # Correct keys matching actual values
        'non_injury_pct': round(float(sev_pcts.get('Unknown', 0.0)), 1),
        'minor_pct':      round(float(sev_pcts.get('Minor',   0.0)), 1),
        'serious_pct':    round(float(sev_pcts.get('Serious', 0.0)), 1),
        'fatal_pct':      round(float(sev_pcts.get('Fatal',   0.0)), 1),
        # Raw counts too — useful for Power BI tooltips
        'non_injury_n':   int(nearby['crashSeverity_clean'].fillna('Unknown').eq('Unknown').sum()),
        'minor_n':        int(nearby['crashSeverity_clean'].eq('Minor').sum()),
        'serious_n':      int(nearby['crashSeverity_clean'].eq('Serious').sum()),
        'fatal_n':        int(nearby['crashSeverity_clean'].eq('Fatal').sum()),
    })

severity_df = pd.DataFrame(rows).sort_values('rank')

print("SEVERITY BREAKDOWN PER TOP 10 KDE HOTSPOT")
print("=" * 75)
print(f"{'#':<4} {'Location':<28} {'Total':>6} {'Non-Inj':>8} {'Minor':>7} {'Serious':>8} {'Fatal':>6}")
print("-" * 75)
for _, r in severity_df.iterrows():
    print(f"  {int(r['rank']):<3} {str(r['crash_location'])[:27]:<28} "
          f"{int(r['total_crashes']):>6} "
          f"{r['non_injury_pct']:>7.1f}% "
          f"{r['minor_pct']:>6.1f}% "
          f"{r['serious_pct']:>7.1f}% "
          f"{r['fatal_pct']:>5.1f}%")
print("=" * 75)

spark.createDataFrame(severity_df).write.format("delta").mode("overwrite") \
     .saveAsTable("gold_top10_severity_breakdown")
print("\nSaved: gold_top10_severity_breakdown")
print("   Columns: rank, crash_location, display_name, total_crashes,")
print("            non_injury_pct, minor_pct, serious_pct, fatal_pct,")
print("            non_injury_n, minor_n, serious_n, fatal_n")
```

```python
# ── 1km grid over Auckland ────────────────────────────────────────────────────
cell_size = 0.010   # ~1km at Auckland latitude

grid_cells = [
    {'geometry': box(lon, lat, lon+cell_size, lat+cell_size),
     'grid_lon': round(lon+cell_size/2, 6),
     'grid_lat': round(lat+cell_size/2, 6)}
    for lon in np.arange(174.45, 175.25, cell_size)
    for lat in np.arange(-37.15, -36.65, cell_size)
]
grid_gdf = gpd.GeoDataFrame(grid_cells, crs="EPSG:4326")
print(f" Grid: {len(grid_gdf):,} cells (~1km × 1km)")

# ── Spatial join — count crashes per cell ─────────────────────────────────────
joined = gpd.sjoin(gdf, grid_gdf, how='left', predicate='within')

grid_counts = joined.groupby('index_right').agg(
    total_crashes  = ('severe_crash', 'count'),   # any column for count
    severe_crashes = ('severe_crash', 'sum'),
    fatal_count    = ('fatalCount',   'sum'),
    serious_count  = ('seriousInjuryCount', 'sum'),
).reset_index().rename(columns={'index_right': 'grid_idx'})

grid_gdf = grid_gdf.merge(grid_counts, left_index=True, right_on='grid_idx', how='left')
for col in ['total_crashes','severe_crashes','fatal_count','serious_count']:
    grid_gdf[col] = grid_gdf[col].fillna(0).astype(int)

# Use total_crashes as Gi* weight — NOT severe_crashes (too sparse to detect clusters)
grid_gdf['gi_weight'] = grid_gdf['total_crashes'].astype(float)

# Active cells only, clean index
grid_active = grid_gdf[grid_gdf['total_crashes'] > 0].copy().reset_index(drop=True)
print(f" Active cells : {len(grid_active):,}")
print(f"   Crashes in grid : {grid_active['total_crashes'].sum():,}  (gdf: {len(gdf):,})")
print(f"   Severe in grid  : {grid_active['severe_crashes'].sum():,}  (gdf: {int(gdf['severe_crash'].sum()):,})")

# ── Distance-band spatial weights (2km radius) ────────────────────────────────
print("\\n Building spatial weights...")
try:
    w_grid = weights.DistanceBand.from_dataframe(
        grid_active, threshold=0.018, silence_warnings=True
    )
    mean_n = np.mean([len(v) for v in w_grid.neighbors.values()])
    print(f" DistanceBand 2km | mean neighbours: {mean_n:.1f}")
except Exception as e:
    print(f"  DistanceBand failed ({e}) — falling back to KNN k=15")
    w_grid = weights.KNN.from_dataframe(grid_active, k=15)

# ── Gi* computation ───────────────────────────────────────────────────────────
print(" Computing Gi* (999 permutations)...")
gi_grid = G_Local(grid_active['gi_weight'].values, w_grid, permutations=999, seed=42)

# Assign results immediately — must not skip these two lines
grid_active['gi_star_z'] = gi_grid.Zs
grid_active['gi_star_p'] = gi_grid.p_sim
print(f" Gi* complete | Z: {grid_active['gi_star_z'].min():.2f} to {grid_active['gi_star_z'].max():.2f}")

# ── Tiered classification — no broad p<0.05 catch-all ────────────────────────
# WRONG pattern (original): has (p<0.05) as 3rd condition which catches everything
# CORRECT: use only z-score thresholds at three confidence levels
grid_active['hotspot_classification'] = np.select(
    [
        (grid_active['gi_star_z'] > 3.29) & (grid_active['gi_star_p'] < 0.001),
        (grid_active['gi_star_z'] > 2.58) & (grid_active['gi_star_p'] < 0.01),
        (grid_active['gi_star_z'] > 1.96) & (grid_active['gi_star_p'] < 0.05),
        (grid_active['gi_star_z'] < -1.96) & (grid_active['gi_star_p'] < 0.05),
    ],
    ['Hot Spot 99.9%', 'Hot Spot 99%', 'Hot Spot 95%', 'Cold Spot 95%'],
    default='Not Significant'
)
print("\\nHOTSPOT RESULTS:")
print(grid_active['hotspot_classification'].value_counts())

# ── Coverage — primary metric: total crashes; secondary: severe ───────────────
hot          = grid_active['hotspot_classification'].str.startswith('Hot Spot')
crashes_hs   = int(grid_active.loc[hot, 'total_crashes'].sum())
severe_hs    = int(grid_active.loc[hot, 'severe_crashes'].sum())
fatal_hs     = int(grid_active.loc[hot, 'fatal_count'].sum())
total_cr     = int(grid_active['total_crashes'].sum())
total_sv     = int(grid_active['severe_crashes'].sum())
total_ft     = int(grid_active['fatal_count'].sum())
crash_cov    = crashes_hs / total_cr * 100
severe_cov   = severe_hs  / total_sv * 100

print(f"\\nCOVERAGE SUMMARY")
print(f"  Total crash coverage : {crash_cov:.1f}%  ({crashes_hs:,}/{total_cr:,})")
print(f"  Severe crash coverage: {severe_cov:.1f}%  ({severe_hs:,}/{total_sv:,})")
print(f"  Fatal coverage       : {fatal_hs}/{total_ft}  ({fatal_hs/total_ft*100:.1f}%)")
print("\\nObjective 1 MET" if crash_cov > 50 else "  Below 50% — run Cell 7 auto-scan")

# ── Save ──────────────────────────────────────────────────────────────────────
spark.createDataFrame(
    grid_active[['grid_idx','grid_lat','grid_lon','total_crashes','severe_crashes',
                 'fatal_count','serious_count','gi_weight',
                 'gi_star_z','gi_star_p','hotspot_classification']]
).write.format("delta").mode("overwrite").saveAsTable("gold_spatial_hotspots")
print("Saved: gold_spatial_hotspots")
```
